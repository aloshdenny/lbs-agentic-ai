# Agent Memory: A Long Conversation

Chapter 01 called memory one of the four ingredients of an agent: *what's happened so far in this run, carried forward.*

Here's the whole idea in three cells: install the SDK, create one client, then keep a growing list of messages and hand the entire list back to the model every turn. That list **is** the memory — there's no separate "memory system" underneath it.

```python
%pip install -q groq
```

```python
from groq import Groq

GROQ_API_KEY = ""
MODEL = "openai/gpt-oss-20b"  # swap this one line to try a different Groq model

client = Groq(api_key=GROQ_API_KEY)
```

```python
messages = [{"role": "system", "content": "You are a helpful, concise assistant."}]

while True:
    user_input = input("You: ")
    if user_input.strip().lower() in ("exit", "quit"):
        break

    messages.append({"role": "user", "content": user_input})

    response = client.chat.completions.create(
        model=MODEL,
        messages=messages,   # the entire conversation goes back in, every single turn
        temperature=0.5,
    )
    reply = response.choices[0].message.content
    print("Assistant:", reply)

    messages.append({"role": "assistant", "content": reply})   # this is what grows the memory

    # try asking it something, then a few turns later asking "what did I just ask you?"
```
